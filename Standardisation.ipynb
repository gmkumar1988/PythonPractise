```python
import pandas as pd
import seaborn as sns
from sklearn.preprocessing import LabelEncoder, OrdinalEncoder, OneHotEncoder
```

Loading the dataset from seaborn 

```python
df = sns.load_dataset('titanic')
```

```python
df = df[['survived','sex','embarked','class','who']].dropna()
```

```python
df.head()
```

```python
le = LabelEncoder()
```

```python
df['sex_le'] = le.fit_transform(df['sex'])
df[['sex','sex_le']].drop_duplicates()
```

```python
df.head
```

Ordinal Encoding

```python
ordinal_order = [['Third','Second','First']]
oe = OrdinalEncoder(categories=ordinal_order)
df['class_ordinal'] = oe.fit_transform(df[['class']])
df[['class','class_ordinal']].drop_duplicates()
```

```python
edu_df = pd.DataFrame({
    'education':['High School','Bachelors','Masters','Phd']})
    ### Define Manual Mapping
education_map = {
    'High School': 0,
    'Bachelors':1,
    'Masters':2,
    'Phd':3 }
```

```python
edu_df['education_encoded'] = edu_df['education'].map(education_map)
```

```python
edu_df
```

```python
gen_df = pd.DataFrame ({
    'gender':['Male','Female','Female','Male']
})
```

```python
gen_df['gender_male'] = (gen_df['gender']== 'Male').astype(int)
gen_df['gender_female'] = (gen_df['gender'] == 'Female').astype(int)
```
