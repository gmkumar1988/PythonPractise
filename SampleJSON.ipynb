```python
import json
from dataclasses import dataclass

@dataclass
class Applicant:
    id: int
    name: str
    income: float
    approved: bool

try:
    with open("loans.json") as f:
        raw = json.load(f)
except FileNotFoundError:
    raw = [
        {"id": 1, "name": "Asha", "income": 52000, "approved": True},
        {"id": 2, "name": "Ben", "income": 38000, "approved": False}
    ]

applicants = [Applicant(**item) for item in raw]

print(applicants[0].name)
print(applicants[0].income)
```
